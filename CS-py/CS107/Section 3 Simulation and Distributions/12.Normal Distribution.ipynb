```python
from scipy.stats import norm
```

```python
print(1 - norm.cdf(1)) # area right of z = 1
print(norm.ppf(.95)) # gives z-score in which the area to the left of it is 95%.
```

```python
avg = 70.5
sd = 3
x = 71.7
z = (x-avg)/sd
per = 0.42
print((1 - norm.cdf(z)) * 100)
print(norm.cdf(z) * 100)
print(norm.ppf(per)*sd+avg)
```
