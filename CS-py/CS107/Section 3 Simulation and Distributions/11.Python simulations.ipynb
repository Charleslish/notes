## Simple simulations in Python  

Using Python, collect 13000 obsercations of rolling a 6-sided die:


```python
data = []
for i in range (13000):
    roll = random.randint(1,6)
    d = {"Roll":roll}
    data.append(d)
df = pd.DataFrame(data)
df
```

Analysis: what is our estimated probability of rolling a number <=4?

```python
df[df["Roll"]<=4].count()/len(df)
```

##  Simulations with Conditional  

simulate 100,000 genetic tests to test someone's relation to Taylor Swift. The test has the following parameters:  
1) There is a 99% probability that an individual related to Taylor Swift will get a positive result ("true positive").  
2) There is a 6% probability that an individual NOT related to Taylor Swift will get a positive result ("false positive").  
3) About 1% of the world population is related to Taylor Swift.  

```python
data = []
for i in range(100000):
    related = random.randint(1,100)
    test = random.randint(1,100)
    if related == 1:
        if test <= 99:
            result = '+'
        else:
            result = '-'
    else:
        if test <= 6:
            result = '+'
        else:
            result = '-'
    d={"related":related, "test":test, "result":result}
    data.append(d)
df = pd.DataFrame(data)
df
```
