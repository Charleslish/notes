```python
import random
```

## Generating Random Numbers in Python  


generate a random number in the **INCLUSIVE** range 1-6

```python
random.randint(1,6)
```

simulate a coin flip

```python
random.randint(0,1)
```

generating random values in a list

```python
random.choice(["Heads","Tails"])
```

it's ok to list the same outcome more than once

```python
random.choice(["Heads","Heads","Heads","Tails","Tails"])
```

## Control flow loop  

repeat code

```python
j = 0
for i in range (5):
    j = j + 1
    print(j)
j
```
