## Grouping Data in Python  

aggregate groups with .groupby(column name), followed by the aggregation function agg(type), where type can be "count", "sum", "mean", etc  
reset_index() makes the dataframe cleaner

```python
df.groupby("A-").agg("count").reset_index()
```

```python
df.groupby("Number").agg("count").reset_index()
```

using numeric_only=True option avoids adding up strings

```python
df.groupby("Number").agg("sum",numeric_only=True).reset_index()
```
