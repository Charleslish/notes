## Data Visualization  

histograms: frequency, density


`df.plot.hist(...)` for histograms  
`df.plot.hist(..., density=True)` for density histograms  
`df.plot.bar(...)` for vertical bar charts  
`df.plot.barh(...)` for horizontal bar charts  
`df.plot.line(...)` for line charts  
`df.plot.scatter(...)` for scatter plots  
`df.plot.pie(...)` for pie charts  
`df.plot.box(...)` for box plots  
`title="..."` for adding titles  
`grid=True` for adding grids  
`yticks=[0.5,1.0]` for setting y-axis ticks   
`xlabel="..."` for adding x-axis label  
`ylabel="..."` for adding y-axis label  
`figsize=(10,5)` for setting figure size  
`color="..."` for setting color  
`edgecolor="..."` for setting edge color  
`legend=True` for adding legend  
`bins=10` for setting number of bins for histograms specifically  

quartile:  

true index location = (n-1) * quartile percentage  

linear interpolated value = low + (high - low) * quartile percentage
