### Row selection

We can call some functions to select rows of the dataframe.


Select 5 random courses offered this semester

```python
df.sample(5)
```

select first 10 courses that are listed in the dataframe

```python
df.head(10)
```

select 7 courses with the largest course number

```python
df.nlargest(7, 'Number')
```

Summary: these are some common operations that we can perform on a DataFrame and their parameters  

`df.head(n)` select the first n rows of the DataFrame  
`df.tail(n)` select the last n rows of the DataFrame  
`df.nlargest(n,'value')` select the n rows with the largest 'value' value  
`df.nsmallest(n,'value')` select the n rows with the smallest 'value' value  
`df.sample(n=10)` randomly select n rows from the DataFrame  
`df.sample(frac=0.1)` randomly select fraction of rows from the DataFrame  
`df.iloc[10:20]` select the rows from the 10th to the 19th row of the DataFrame  
`df.drop_duplicates()` remove the rows with duplicate values  
`df[condition]` select the rows that satisfy the condition
