## Lists  

A list is also called an array.

create an empty list named 'a'  

lists can hold elements of different types, of the same value, and of any type  

```python
a = []
b = [3, 'Python', 2, 3, [a, 2]]
```

Accessing data in a list by calling the index

```python
b[2]
```

```python
b[4]
```

appending new data to the end of a list by append()

```python
b.append(4)
b
```

## Functions  

General Syntax: \<function name\>(\<parameter lists\>)

```python
def add(a,b):
    return a + b
add(4, 9)
```

if we set default values, they will run it with those values if no value is provided in the parameter list

```python
def subtract(number, amount = 10):
    return number - amount
subtract(4)
```

```python
subtract(4,12)
```
