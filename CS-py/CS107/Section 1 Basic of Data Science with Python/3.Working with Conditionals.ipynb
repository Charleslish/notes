## Working with conditionals

Some logic in Python (and pandas):  
`<` less than  
`>` greater than  
`<=` less than or equal to  
`>=` greater than or equal to  
`==` equal to  
`!=` not equal to  
`df.column.isin(values1)` return true if the value in the column is in values1  
`pd.isnull(obj)` is NaN  
`pd.notnull(obj)` is not NaN  
`&` and  
`|` or  
`~` not  
`^` exclusive or (xor)  
`df.any()` return true if any value in the DataFrame is true  
`df.all()` return true if all values in the DataFrame are true  


```python
df[df.Number > 500]
```

The following code uses a & to join 2 conditionals together  

it must use () because it would otherwise take df.Number as its other conditional

```python
df[(df.Number >= 100) & (df.Number < 200)]
```
