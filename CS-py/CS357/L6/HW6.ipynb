```python
import math

x = 3

# f(x) = 5*x**3 + 4*x**2 + 2*x + 6
# Taylor 二阶展开（在 x=0 附近）
# f(x) ≈ f(0) + f'(0)x + f''(0)x^2/2
# f(0) = 6, f'(0) = -2, f''(0) = 4
#  f(x) ≈ 6 - 2x + 4x^2/2
# 精确值

exact = 7*x**3 + 5*x**2 + 3*x + 4
approx = 4 + 10*(x-3) - 21*(x-3)**2



# 绝对误差
error = -(exact - approx)

print("Taylor approximation:", approx)
print("Exact value:", exact)
print("Absolute error:", error)
```
