# 3) Catastrophic Cancellation

Let's make two numbers with very similar magnitude:

```python
x = 1.48234
y = 1.48235
```

Now let's compute their difference in double precision:

```python
x_dbl = np.float64(x)
y_dbl = np.float64(y)
diff_dbl = x_dbl-y_dbl

diff_dbl
```

* What would the correct result be?  
$-10^{-5}$
* What has happened here?  
The small error occurs because decimal numbers cannot always be represented exactly in binary floating-point format. Subtracting two nearly equal numbers makes the rounding error more noticeable due to cancellation.

-------------
