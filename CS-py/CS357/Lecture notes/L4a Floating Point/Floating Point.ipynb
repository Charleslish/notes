# Floating-point

## (Unsigned) Fixed-point representation

The numbers are stored with a fixed number of bits for the integer part and a fixed number of bits for the fractional part.  
Suppose we have 8 bits to store a real number, where 5 bits store the integer part and 3 bits store the fractional part:  
$(10111.011)_2 = 23.375$  
smallest number: $(00000.001)_2 = 0.0625$  
largest number: $(11111.111)_2 = 31.9375$  

Suppose we have 64 bits to store a real number, where 32 bits store the integer part and 32 bits store the fractional part:
$$
(a_{31}a_{30}\dots a_{0}.b_{1}b_{2}\dots b_{32})_2 = \sum_{k=0}^{31} a_{k}2^{k} + \sum_{k=1}^{32} b_{k}2^{-k}  
= a_{31}2^{31} + a_{30}2^{30} + \dots + a_{0}2^0 + b_{1}2^{-1} + b_{2}2^{-2} + \dots + b_{32}2^{-32}
$$
smallest number: $(00000000000000000000000000000000.00000000000000000000000000000001)_2 = 2^{-63} \approx 10^{-9}$  
largest number: $(11111111111111111111111111111111.11111111111111111111111111111111)_2 = 2^{63} - 2^{-63} \approx 10^{9}$

How can we decide where to locate the binary point?  
moving the point changes the bits for integer part and fractional part  
  
**Range**: difference between the largest and smallest numbers possible.  
&emsp;&emsp;More bits for the integer part ⟶ increase range  
  
**Precision**: smallest possible difference between any two numbers  
&emsp;&emsp;More bits for the fractional part ⟶ increase precision  

Let the binary point “float” for a trade off between range and precision.

## Scientific Notation

In scientific notation, a number can be expressed in the form
$$
x = \pm r \times 10^m
$$

where $r$ is a coefficient in the range $1 \leq r < 10$, and $m$ is the exponent.

examples:  
$
1165.7 = 1.1657 \times 10^3 \\
0.0004728 = 4.728 \times 10^{-4}
$

## Floating-point numbers
A floating-point number can represent numbers of different order of
magnitude (very large and very small) with the same number of fixed
digits.  
In general, in the binary system, a floating number can be expressed as
$$
x = \pm q \times 2^m
$$

where $q$ is the significand, normally a fractional value in the range $[1.0,2.0) $, and $m$ is the exponent.

**Numerical Form**:
$$
    x = \pm r \times 10^m = \pm b_0.b_1b_2b_3...b_n \times 2^m
$$

$b_i \in \{0,1\}$  
$b_0$ is the **sign bit** ($0$ for positive, $1$ for negative)  
$b_1b_2b_3...b_n$ is the **fractional part**  
**Exponent** range: $m \in [L,U]$  
**Precision**: $\boxed{p = n + 1}$  


## Converting floating points

Convert $(39.6875)_{10} = (100111.1011)_{2}$ into floating point representation:
&emsp;&emsp;$1.001111011 \times 2^5$

## Normalized floating-point numbers

Normalized floating point numbers are expressed as 
$$
    x = \pm 1.b_1b_2b_3...b_n \times 2^m = \pm 1.f \times 2^m
$$
where $f = b_1b_2b_3...b_n$ is the **fractional part** of the significand, $b_i \in \{0,1\}$ and $m \in [L,U]$ is the **exponent**.  
The 1 is the "**hidden bit**" and is always 1.  
We "gain" an extra bit of precision by using the hidden bit: $p = n + 1$  

**Smallest positive normalized FP number:** $1.\underbrace{00\cdots0}_{n\text{ zeros}} \times 2^L = \boxed{2^L}$

**Largest positive normalized FP number:** $1.\underbrace{11\cdots1}_{n\text{ zeros}} \times 2^U = \boxed{2^{U+1} \times (1-2^{-p})}$  


```python
L = -7
U = 6
n = 3
p = n + 1
small = 2**L
big = 2**(U+1) * (1-2**(-p))
print(f"smallest: {small}")
print(f"largest: {big}")
```

## Machine epsilon
**Machine epsilon ($\epsilon _m$):** is defined as the distance (gap) between 1 and the
next larger floating point number  
In general, it is defined as $\boxed{\epsilon _m = 2^{-n}}$  
For $\pm 1.b_1b_2 \times 2^m$ for $m \in [-4,4]$ and $b_i \in \{0,1\}$,  
$\epsilon _m = 2^{-2} = 0.25$
