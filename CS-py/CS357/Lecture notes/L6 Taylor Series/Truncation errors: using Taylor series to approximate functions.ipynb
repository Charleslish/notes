# Truncation errors: using Taylor series to approximate functions  

The Taylor series of $f(x)$ about $x_0$ is given by:
$$
f(x) = f(x_0) + f'(x_0)(x - x_0) + \frac{f''(x_0)(x - x_0)^2}{2!} + \frac{f'''(x_0)(x - x_0)^3}{3!} + \cdots
$$
$$
f(x) = \sum_{i=0}^\infty \frac{f^{(i)}(x_0)}{i!}(x - x_0)^i
$$

## Taylor Series – what is the error?

We cannot sum infinite number of terms, and therefore we have to truncate.  
How big is the error caused by truncation? Let’s write $h = x - x_0$
$$
f(x_0 + h) = f(x_0) + f'(x_0)h + \frac{f''(x_0)}{2!}h^2 + \frac{f'''(x_0)}{3!}h^3 + \cdots
$$
$$
f(x_0 + h) = \underbrace{\sum_{i=0}^n \frac{f^{(i)}(x_0)}{i!}h^i}_{t_n(x)} + \underbrace{\sum_{i=n+1}^{\infty} \frac{f^{(i)}(x_0 + h)}{i!}h^i}_{\text{error}}
$$

$$
|\text{error}| = | \sum_{i=n+1}^{\infty} \frac{f^{(i)}(x_0 + h)}{i!}h^i | = \frac{|f^{(n+1)}(x_0 + h)|}{(n+1)!}h^{n+1} + \frac{|f^{(n+2)}(x_0 + h)|}{(n+2)!}h^{n+2} + \cdots
$$
$$
\Rightarrow |\text{error}| \leq \frac{|f^{(n+1)}(x_0 + h)|}{(n+1)!}h^{n+1} \leq |M|h^{n+1}|
$$
$$
\Rightarrow |\text{error}| = O(h^{n+1})
$$

$e_1 \propto h_1^{n+1}$, $e_2 \propto h_2^{n+1}$  

$\Rightarrow \frac{e_1}{e_2} = \frac{h_1^{n+1}}{h_2^{n+1}} = (\frac{h_1}{h_2})^{n+1}$  

$\Rightarrow e_2 = (\frac{h_2}{h_1})^{n+1} e_1$

A function $f(x)$ has the following Taylor series expansion $t(x)$ about $x = 3$:  
$$t(x) = \frac{7}{8} + \frac{7}{2}(x-3) + \frac{5}{3}(x-3)^3 + \frac{12}{5}(x-3)^5$$
Approximate the derivative of the function at $x = 3.2$ using the Taylor approximation of degree 2.

<details>
<summary>Solution</summary>

