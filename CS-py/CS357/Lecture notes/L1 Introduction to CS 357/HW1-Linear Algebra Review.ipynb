### 2. Basis of a Vector Space

Let 
$\mathbf{v} = \begin{bmatrix} 11 & 2 \end{bmatrix}^T.$
For which of the following $\mathbf{w}$ does the set $\{\mathbf{v}, \mathbf{w}\}$ not span $\mathbb{R}^2?$

$$
a) \mathbf{w} = \begin{bmatrix} -2 & 11 \end{bmatrix}^T \quad
b) \mathbf{w} = \begin{bmatrix} -11 & -2 \end{bmatrix}^T \quad  
c) \mathbf{w} = \begin{bmatrix} 77 & 14 \end{bmatrix}^T \quad
d) \mathbf{w} = \begin{bmatrix} 3 & 11 \end{bmatrix}^T \quad
e) \mathbf{w} = \begin{bmatrix} 2 & 3 \end{bmatrix}^T \quad
$$

<details>
<summary>Solution</summary>

$$
b) \mathbf{w} = \begin{bmatrix} -11 & -2 \end{bmatrix}^T \quad
c) \mathbf{w} = \begin{bmatrix} 77 & 14 \end{bmatrix}^T \quad
$$

### 14. Augmented Matrix and number of solutions

number of columns in coefficient matrix - number of pivot columns = number of free variables
infinite solutions: number of free variables > 0 and no 0 rows
only 1 solution: number of pivot columns = number of rows
no solution: number of free variables < 0, or number of pivot columns < number of rows

### 17. Eigenvalues and Eigenvectors

eigenvalue: scalar $\lambda$ such that $A\vec{x} = \lambda\vec{x}$  
eigenvector: non-zero vector $\vec{x}$ such that $A\vec{x} = \lambda\vec{x}$  
use the characteristic equation to find eigenvalues: $det(A - \lambda I) = 0$  
or use the determinant of $A$: $det(A) = \lambda_n\lambda_{n-1}\cdots\lambda_1\lambda_0$
