# Floating point and finite number representation

What will the following code snippet do?

```python
from time import sleep

x = 0.0

while x <= 1.0:
    x += 0.1
    print(repr(x))
    
    sleep(0.1)
```

How do you prevent that from happening?

```python
a = 0.1 + 0.2
b = 0.3
epsilon = 1e-9 

print(a == b) 
print(abs(a - b) < epsilon) 
```

```python
from decimal import Decimal, getcontext
from time import sleep

getcontext().prec = 20

x = Decimal('0.0') 

while x <= Decimal('1.0'):
    x += Decimal('0.1')
    print(repr(x))
    sleep(0.1)
```

# Complexity: Timing Matrix Matrix Multiplication

```python
from time import process_time
import numpy as np
```

```python
n = 2000
A = np.random.randn(n,n)
B = np.random.randn(n,n)

t = process_time()  # store the time
C = A @ B # matrix multiplication
t = process_time() - t
print(t)
```

```python
n = 2000
a = 2
A = np.random.randn(a*n,a*n)
B = np.random.randn(a*n,a*n)

t2 = process_time()  # store the time
C = A @ B # matrix multiplication
t2 = process_time() - t2
print(t2)
```

# Plotting time versus matrix dimension

```python
import numpy as np
import matplotlib.pyplot as pt
%matplotlib inline
from time import process_time
```

```python
def get_solve_time(n):
    A = np.random.randn(n, n)
    B = np.random.randn(n, n)
    
    t_start = process_time()
    A @ B
    t_stop = process_time()
    
    return t_stop-t_start
```

```python
# n_values = np.array([100,500,1000,1500,2000,2500])
# print(n_values)
# # if you are running this notebook locally, you can try running the next cell for incresing matrix sizes
n_values = np.array([100,1000,2000,3000,4000,5000,6000,8000,10000])
print(n_values)
```

```python
times = []
for n in n_values:
    newtime = get_solve_time(n)
    times.append(newtime)
```

```python
pt.loglog(n_values, times,'o-')
pt.xlabel('n')
pt.ylabel('time')
pt.grid()
```

* Can we predict individual values?  

It's very hard to accurately predict individual values. However, we can predict a rough prediction of the individual values based on the overall behavior of the function.  

* What does the overall behavior look like?  

The overall behavior of the function is to map the input values to the output values. The mapping is usually non-linear. In this example, the function is approximately O(n^3) for matrix multiplication.  

* How could we determine the "underlying" function?  

We can determine the underlying function by analyzing the behavior of the function for different input values. We can also use mathematical techniques to determine the underlying function. In this example, we can use the definition of matrix multiplication to determine the underlying function.
