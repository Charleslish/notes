Consider the following vector:

$$
{\bf v} = \begin{bmatrix}  -1\\  0\\  3\\  4\\\end{bmatrix}
$$

Calculate the 2, 3, 5 and $\infty$ norms.

```python
import numpy as np
import numpy.linalg as la

n = np.array([-1,0,3,4])
print(la.norm(n, 2))
print(la.norm(n, 3))
print(la.norm(n, 5))
print(la.norm(n, np.inf))
```

Given $x_{approx}$ = $\begin{bmatrix} 3 \\ 3 \end{bmatrix}$, 2-norm absolute error = $4\sqrt{2}$, and $x_{real}$ = $\begin{bmatrix} a \\ a \end{bmatrix}$ for some $a \in \mathbb{R}$ .

Find the largest possible $a$ (in magnitude).



<details>
<summary>
Solution
</summary>

magnitude of error = $\|\text{true - approximate}\|$  
$ \sqrt{(a-3)^2 + (a-3)^2} = 4\sqrt{2} \Rightarrow a = 7$
