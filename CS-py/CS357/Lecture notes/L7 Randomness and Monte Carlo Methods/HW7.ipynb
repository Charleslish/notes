Your favorite shopping site is offering a lucky draw for every item you purchase and you want to estimate the chance of winning a $10 credit per draw using the monte carlo method.  
Assuming that the chances of winning is identical each time, after trying for 800 times, the error between the estimated winning probability and the pre-defined winning probability in the website's setting is $10^{-2}$.
If you try the drawing for a total of $8 \times 10^6$ times, what would the estimation error approximately be?  
The error for Monte-Carlo Methods is $\boxed{O(\frac{1}{\sqrt{N}})}$
