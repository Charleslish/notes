# IEEE 754 Standard

Numerical form:
$$
    x = \pm 1.f \times 2^m
$$
where $f$ is the **fractional part** of the significand, $m \in [L,U]$ is the **exponent**.  
$c = m + shift$ is an unsigned int  
Represent in memory:  
IEEE-754 Single precision (32 bits): x = sign bit + 8 bits for exponent + 23 bits for significand  
IEEE-754 Single precision (64 bits): x = sign bit + 11 bits for exponent + 52 bits for significand  
Generally, x = {0,1} + $(c)_2$ + $f$  


## IEEE-754 single precision (32 bits)
$$
    x = (-1)^s \times 1.f \times 2^m
$$  
where $m = c - shift$, c has 8 bits, $2^8 = 256$  
reserve 0 and 255 for special cases  
Therefore, $\boxed{1 \leq c \leq 254}$  
set $shift = 254 \div 2 = 127$  
$\Rightarrow \boxed{-126 \leq m \leq 127}$  
$
s =
\begin{cases}
0, & x\geq 0,\\
1, & x<0.
\end{cases}
$  
$\boxed{n = 23}$, 
$p = n + 1 = 24$

Example: $(67.125)_{10} = (1000011.001)_{2} = 1.000011001 \times 2^6$  
$\Rightarrow s = 0, m = 6, c = m + 127 = 133 = (\underbrace{10000101}_{8\text{ bits}})_2$  
$f = \underbrace{000011001000\cdots0}_{23\text{ bits}}$  
$x = \underbrace{0}_{1\text{ bit}} \underbrace{10000101}_{8\text{ bits}} \underbrace{0000110010\cdots0}_{23\text{ bits}}$

**Machine epsilon ($\epsilon _m$):** is defined as the distance (gap) between 1 and the
next larger floating point number  
$\boxed{\epsilon _m = 2^{-n} = 2^{-23}}$  
**Smallest positive normalized FP number:** $2^L = 2^{-126} \approx 10^{-38}$

**Largest positive normalized FP number:** $2^{U+1} \times (1-2^{-p}) = 2^{128} \times (1-2^{-24}) \approx 10^{38}$  

## IEEE-754 single precision (64 bits)
$$
    x = (-1)^s \times 1.f \times 2^m
$$  
where $m = c - shift$, c has 11 bits, $2^{11} = 2048$  
reserve 0 and 2047 for special cases  
Therefore, $\boxed{1 \leq c \leq 2046}$  
set $shift = (2046-0) \div 2 = 1023$  
$\Rightarrow \boxed{-1022 \leq m \leq 1023}$  
$
s =
\begin{cases}
0, & x\geq 0,\\
1, & x<0.
\end{cases}
$  
$\boxed{n = 52}$, 
$p = n + 1 = 53$

**Machine epsilon ($\epsilon _m$):** is defined as the distance (gap) between 1 and the
next larger floating point number  
$\boxed{\epsilon _m = 2^{-n} = 2^{-52} \approx 2.2 \times 10^{-16}}$  
**Smallest positive normalized FP number:** $2^L = 2^{-1022} \approx 2.2 \times 10^{-308}$

**Largest positive normalized FP number:** $2^{U+1} \times (1-2^{-p}) = 2^{1024} \times (1-2^{-53}) \approx 1.8 \times 10^{308}$  

## Special Values
1) **Zero:** $0 = s \underbrace{00\cdots00}_{8,11\text{ bits}} \underbrace{0000\cdots0}_{23,52\text{ bits}}$
2) **Positive Infinity:** $+\infty = (s=1) \underbrace{11\cdots11}_{8,11\text{ bits}} \underbrace{0000\cdots0}_{23,52\text{ bits}}$
3) **Negative Infinity:** $-\infty = (s=0) \underbrace{11\cdots11}_{8,11\text{ bits}} \underbrace{0000\cdots0}_{23,52\text{ bits}}$
4) **NaN,(results from operations with undefined results):** $NaN = s \underbrace{11\cdots11}_{8,11\text{ bits}} \underbrace{anything \neq 00\cdots0}_{23,52\text{ bits}}$
5) **Subnormal Number:** $x = s \underbrace{00\cdots00}_{8,11\text{ bits}} \underbrace{anything \neq 00\cdots0}_{23,52\text{ bits}}$


## Subnormal (or denormalized) numbers

Noticeable gap around zero, present in any floating system, due to normalization  
1) The smallest possible significand is $1.00\cdots0$  
2) The smallest possible exponent is $L$  

Relax the requirement of normalization, and allow the leading digit to be zero, only when the exponent is at its minimum ($m = L$)
$$
    x = (-1)^s \times 0.f \times 2^m
$$  

32-bits: set $m = L = -126$  
**Smallest positive subnormalized FP number:** $0.\underbrace{00\cdots01}_{23\text{ bits}} \times 2^L = 2^{-23} \times 2^{-126} = 2^{-149} \approx 1.4 \times 10^{-45}$  
**Largest positive subnormalized FP number:** $0.\underbrace{11\cdots11}_{23\text{ bits}} \times 2^L = (1 - 2^{-23}) \times 2^{-126} = 2^{-126} - 2^{-149}$  
64-bits: set $m = L = -1022$  
**Smallest positive subnormalized FP number:** $0.\underbrace{00\cdots01}_{23\text{ bits}} \times 2^L = 2^{-52} \times 2^{-1022} = 2^{-1074} \approx 10^{-324}$  
**Largest positive subnormalized FP number:** $0.\underbrace{11\cdots11}_{23\text{ bits}} \times 2^L = (1 - 2^{-52}) \times 2^{-1022} = 2^{-1022} - 2^{-1074}$


PROS: More gradual underflow to zero  
CONS: - Computations with subnormal numbers are often slow;
-Loss of precision
