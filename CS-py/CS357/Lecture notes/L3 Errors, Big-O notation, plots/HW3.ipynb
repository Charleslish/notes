The current flowing through a wire(*$I$*) is 1 ampere with a relative error of 6%.
The resistance of the wire(*$R$*) is 20 ohms with a relative error of 3%.
What is the relative error in the voltage across the wire knowing the voltage is given by, 
$$V = IR$$


<details>
<summary>Solution</summary>

The relative error in the voltage across the wire is given by,
$$\frac{\Delta V}{V} \approx \frac{\Delta I}{I} + \frac{\Delta R}{R} = 9\%$$

</details>
